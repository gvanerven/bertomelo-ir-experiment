```python
from datasets import load_dataset
from transformers import AutoTokenizer
import pandas as pd
```

```python
model_id = "unb-labia/BERTomelo-ModernBERT-Base-v1"
tokenizer = AutoTokenizer.from_pretrained(model_id)
```

```python
quati_1M = load_dataset("unicamp-dl/quati", trust_remote_code=True)
quati_1M_qrels = load_dataset("unicamp-dl/quati", "quati_1M_qrels", trust_remote_code=True)
quati_all_topics = load_dataset("unicamp-dl/quati", "quati_all_topics", trust_remote_code=True)
```

```python
quati_1M
```

```python
quati_1M['quati_1M_passages']['passage']

qtds_quati = []
for passage in quati_1M['quati_1M_passages']['passage']:
    qtds_quati.append({"palavras": len(passage.split()), "tokens": tokenizer(passage, return_tensors="pt")['input_ids'].shape[1]})

```

```python
df_quati = pd.DataFrame(qtds_quati)
df_quati
```
